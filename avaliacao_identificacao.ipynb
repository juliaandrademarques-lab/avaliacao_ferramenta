## Matriz de confusão

Para avaliar o desempenho da ferramenta na **identificação da presença ou ausência dos documentos**, os resultados produzidos pela ferramenta são comparados aos resultados de referência, considerados como o resultado correto para a avaliação.

A **matriz de confusão** permite organizar os resultados dessa comparação em quatro categorias:

- **Verdadeiro Positivo (VP):** o documento está **Presente** segundo a referência e a ferramenta também o identifica como **Presente**.
- **Verdadeiro Negativo (VN):** o documento está **Ausente** segundo a referência e a ferramenta também o identifica como **Ausente**.
- **Falso Positivo (FP):** o documento está **Ausente** segundo a referência, mas a ferramenta o identifica como **Presente**.
- **Falso Negativo (FN):** o documento está **Presente** segundo a referência, mas a ferramenta o identifica como **Ausente**.

Nesta avaliação, **"Presente" é considerada a classe positiva**, enquanto **"Ausente" é considerada a classe negativa**.

A partir dessas quatro categorias, são calculadas as métricas de desempenho da ferramenta, como **acurácia, sensibilidade, especificidade, precisão e F1-score**.

Os valores de VP, VN, FP e FN são obtidos diretamente dos dados da planilha, de forma automática, para que os resultados sejam atualizados caso a base seja modificada.

```python
pd.crosstab(
    df["Resultado de referência"],
    df["Resultado da ferramenta"],
    margins=True
)
```

```python
vp = ((df["Resultado de referência"] == "Presente") & 
      (df["Resultado da ferramenta"] == "Presente")).sum()

vn = ((df["Resultado de referência"] == "Ausente") & 
      (df["Resultado da ferramenta"] == "Ausente")).sum()

fp = ((df["Resultado de referência"] == "Ausente") & 
      (df["Resultado da ferramenta"] == "Presente")).sum()

fn = ((df["Resultado de referência"] == "Presente") & 
      (df["Resultado da ferramenta"] == "Ausente")).sum()

print(f"Verdadeiros Positivos (VP): {vp}")
print(f"Verdadeiros Negativos (VN): {vn}")
print(f"Falsos Positivos (FP): {fp}")
print(f"Falsos Negativos (FN): {fn}")
```

## Acurácia

A **acurácia** representa a proporção de identificações realizadas corretamente pela ferramenta em relação ao total de documentos avaliados.

Nesse caso, considera-se como acerto tanto quando a ferramenta identifica corretamente um documento como **Presente** quanto quando identifica corretamente um documento como **Ausente**.

A acurácia é calculada pela seguinte fórmula:

$$
\text{Acurácia} = \frac{VP + VN}{VP + VN + FP + FN}
$$

Em que:

- **VP** = Verdadeiros Positivos;
- **VN** = Verdadeiros Negativos;
- **FP** = Falsos Positivos;
- **FN** = Falsos Negativos.

Quanto mais próxima de 100% a acurácia estiver, maior é a proporção de identificações corretas realizadas pela ferramenta.

```python
acuracia = (vp + vn) / (vp + vn + fp + fn)

print(f"Acurácia: {acuracia:.2%}")
```

## Sensibilidade

A **sensibilidade** mede a capacidade da ferramenta de identificar corretamente os documentos que estão **Presentes** no processo.

Em outras palavras, indica a proporção de documentos que estavam presentes segundo o resultado de referência e que também foram identificados como presentes pela ferramenta.

A sensibilidade é calculada pela seguinte fórmula:

$$
\text{Sensibilidade} = \frac{VP}{VP + FN}
$$

Em que:

- **VP** = Verdadeiros Positivos;
- **FN** = Falsos Negativos.

Uma sensibilidade elevada indica que a ferramenta apresenta menor probabilidade de **deixar de identificar documentos que estão presentes**.

```python
sensibilidade = vp / (vp + fn)

print(f"Sensibilidade: {sensibilidade:.2%}")
```

## Especificidade

A **especificidade** mede a capacidade da ferramenta de identificar corretamente os documentos que estão **Ausentes** no processo.

Em outras palavras, indica a proporção de documentos que estavam ausentes segundo o resultado de referência e que também foram identificados como ausentes pela ferramenta.

A especificidade é calculada pela seguinte fórmula:

$$
\text{Especificidade} = \frac{VN}{VN + FP}
$$

Em que:

- **VN** = Verdadeiros Negativos;
- **FP** = Falsos Positivos.

Uma especificidade elevada indica que a ferramenta apresenta menor probabilidade de **indicar como presente um documento que, na realidade, está ausente**.

```python
especificidade = vn / (vn + fp)

print(f"Especificidade: {especificidade:.2%}")
```

## Precisão

A **precisão** mede a proporção de identificações como **Presente** realizadas corretamente pela ferramenta.

Em outras palavras, indica, entre todos os documentos que a ferramenta identificou como presentes, quantos estavam realmente presentes segundo o resultado de referência.

A precisão é calculada pela seguinte fórmula:

$$
\text{Precisão} = \frac{VP}{VP + FP}
$$

Em que:

- **VP** = Verdadeiros Positivos;
- **FP** = Falsos Positivos.

Uma precisão elevada indica que, quando a ferramenta identifica um documento como **Presente**, há maior probabilidade de essa identificação estar correta.

```python
precisao = vp / (vp + fp)

print(f"Precisão: {precisao:.2%}")
```

## F1-score

O **F1-score** é uma medida que combina **precisão** e **sensibilidade** em um único indicador.

Ele é calculado por meio da média harmônica entre precisão e sensibilidade, atribuindo o mesmo peso às duas métricas.

A fórmula é:

$$
F1 = 2 \times \frac{\text{Precisão} \times \text{Sensibilidade}}
{\text{Precisão} + \text{Sensibilidade}}
$$

O F1-score varia de 0 a 1, podendo também ser apresentado em percentual. Quanto mais próximo de 100%, maior é o equilíbrio entre a capacidade da ferramenta de identificar corretamente os documentos presentes e a proporção de suas identificações positivas que estão corretas.

```python
f1 = 2 * (precisao * sensibilidade) / (precisao + sensibilidade)

print(f"F1-score: {f1:.2%}")
```

## Síntese das métricas de desempenho

As métricas calculadas são consolidadas a seguir para facilitar a visualização do desempenho da ferramenta na identificação da presença ou ausência dos documentos.

Os resultados são calculados automaticamente a partir dos dados da planilha e das classificações de Verdadeiros Positivos (VP), Verdadeiros Negativos (VN), Falsos Positivos (FP) e Falsos Negativos (FN).

```python
metricas = pd.DataFrame({
    "Métrica": [
        "Acurácia",
        "Sensibilidade",
        "Especificidade",
        "Precisão",
        "F1-score"
    ],
    "Resultado": [
        acuracia,
        sensibilidade,
        especificidade,
        precisao,
        f1
    ]
})

metricas["Resultado"] = metricas["Resultado"].map(lambda x: f"{x:.2%}")

metricas
```
